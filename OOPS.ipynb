#PART 1: THEORETICAL QUESTION
#1. What is Object-Oriented Programming (OOP)?

-  Object-Oriented Programming (OOP) is a programming paradigm that organizes software design around data, or objects, rather than functions and logic. It models real-world entities using classes and objects, structured around four core pillars:  Encapsulation: Bundling data and methods together.  Inheritance: Reusing code from existing classes.  Polymorphism: Allowing different classes to be treated through a common interface.  Abstraction: Hiding complex implementation details.  
#2. What is a class in OOP?

-  A class is a blueprint or template for creating objects. It defines a blueprint by specifying the attributes (data) and methods (functions) that the objects created from it will possess.  
#3. What is an object in OOP?

-  An object is an instance of a class. When a class is defined, no memory is allocated until an object of that class is created, allowing you to interact with the actual data and behaviors defined by the class.
# 4. What is the difference between abstraction and encapsulation?

-  Abstraction focuses on hiding the internal complexity and showing only the essential features of an object. It answers what the object does rather than how it does it (e.g., using abstract classes).  Encapsulation is the practice of bundling data and methods into a single unit (a class) and restricting direct access to some of the object's components to prevent accidental modification.
# 5. What are dunder methods in Python?

-  Dunder methods (short for Double Underscore methods, also called magic methods) are special built-in methods that start and end with two underscores (e.g., __init__, __str__). They allow you to define custom behaviors for built-in operators or operations, such as object creation, mathematical operations, or string representation.  

#6. Explain the concept of inheritance in OOP.

-  Inheritance is a mechanism that allows a new class (child or derived class) to inherit attributes and methods from an existing class (parent or base class). It promotes code reusability and establishes an "is-a" relationship between classes.  

#7. What is polymorphism in OOP?

-  Polymorphism literally means "many forms". It allows different objects to respond to the same method call in their own unique way. This means a uniform interface can be used to control access to different underlying implementations.
# 8. How is encapsulation achieved in Python?

-  Python achieves encapsulation by using specific naming conventions to restrict access to variables and methods:Protected attributes: Prefixed with a single underscore _variable. It's a hint to developers that the attribute is private, though it can still be accessed.Private attributes: Prefixed with a double underscore __variable. Python uses name mangling to change the internal name of the variable, making it harder to access directly from outside the class.

# 9. What is a constructor in Python?

- A constructor is a special method used to initialize the attributes of an object when it is created. In Python, the constructor is defined using the __init__() dunder method, which runs automatically whenever a new object is in stantiated.

# 10. What are class and static methods in Python?

-  Class Method (@classmethod): Takes cls as its first argument and can access or modify class-level state that applies across all instances.  Static Method (@staticmethod): Does not take self or cls as an argument. It behaves like a regular function but belongs to the class's namespace because it is logically related to the class.  

# 11. What is method overloading in Python?

-  Method overloading is the ability to define multiple methods with the same name but different parameters. Python does not natively support traditional method overloading based on parameter types or counts. If you define multiple methods with the same name, Python only recognizes the last one. Overloading is typically mimicked using default arguments or variable-length arguments (*args, kwargs).

# 12. What is method overriding in OOP?

-  Method overriding occurs when a child class provides a specific implementation for a method that is already defined in its parent class. The child class version replaces the parent's implementation for instances of the child class.

# 13. What is a property decorator in Python?

-  The @property decorator is used to turn a class method into a "getter" attribute. It allows you to access a method's return value like a regular attribute without using parentheses (e.g., obj.value instead of obj.value()), enabling cleaner access control and data validation.


# 14. Why is polymorphism important in OOP?


-  Polymorphism is crucial because it allows flexibility and extensibility. It enables you to write code that can process different types of objects dynamically as long as they share a common interface. This means you can add new classes without breaking or modifying existing code.  

#15. What is an abstract class in Python?


-   An abstract class is a template class that cannot be instantiated directly. It must be inherited by other classes, which are required to implement its abstract methods. It is created using Python’s abc module and the @abstractmethod decorator.  

#16. What are the advantages of OOP?

-  Code Reusability: Inheritance avoids redundant code.  Modularity: Classes break down complex systems into manageable, independent pieces.  Easier Maintenance: Code changes are isolated within encapsulated modules, reducing side effects.  Security: Data hiding through encapsulation protects internal states.


#17. What is the difference between a class variable and an instance variable?

-  Class Variable: Shared by all instances of a class. It is defined directly inside the class body, outside any methods.  Instance Variable: Unique to each specific instance of a class. It is defined inside methods (typically __init__) using self.

# 18. What is multiple inheritance in Python?

-   Multiple inheritance occurs when a child class inherits attributes and behaviors from more than one parent class. Python resolves the method execution order using the Method Resolution Order (MRO) algorithm.  


#19. Explain the purpose of __str__ and __repr__ methods in Python.


-  __str__: Used to return a user-friendly, readable string representation of an object, intended for end-users (triggered by print() or str()).  __repr__: Used to return an unambiguous, official string representation of the object, intended for developers and debugging (triggered by repr()). Ideally, it should look like the Python code needed to recreate the object.

#20. What is the significance of the super() function in Python?


-  The super() function returns a proxy object that delegates method calls to a parent or sibling class. It is most commonly used to call the parent class's constructor (super().__init__()) to ensure that base initializations occur correctly without explicitly naming the parent class.  

#21. What is the significance of the __del__ method in Python?

-  The __del__ method is a destructor. It is called automatically by Python's garbage collector right before an object's memory is deleted. It is typically used to perform clean-up operations, like closing open database connections or file handles.  

#22. What is the difference between @staticmethod and @classmethod in Python?

-  @classmethod receives the class object itself (cls) as an implicit first argument, allowing it to modify class state.  @staticmethod receives no implicit first argument (neither self nor cls), acting exactly like an isolated utility function tied to the class namespace.

# 23. How does polymorphism work in Python with inheritance?


-  When a child class inherits from a parent class, it can override any parent methods. When you call that method on an object variable, Python checks the actual object type at runtime and executes the overridden child version rather than the parent version, achieving runtime polymorphism.  

#24. What is method chaining in Python OOP?



-Method chaining is a design pattern where multiple methods are called sequentially on a single object in a single line of code (e.g., object.method1().method2().method3()). This is achieved by having each method return self at the end of its execution block.   
#25. What is the purpose of the __call__ method in Python?

-  The __call__ method allows an instance of a class to be called exactly like a regular function. When implemented, writing object() invokes the behavior defined inside its __call__ method.

```python
#ANSWER 2
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


c = Circle(5)
r = Rectangle(4, 5)
print(f"Circle Area: {c.area():.2f}")
print(f"Rectangle Area: {r.area()}")
```

```python
#ANSWER 3
class Vehicle:
    def __init__(self, v_type):
        self.type = v_type

class Car(Vehicle):
    def __init__(self, v_type, brand):
        super().__init__(v_type)
        self.brand = brand

class ElectricCar(Car):
    def __init__(self, v_type, brand, battery):
        super().__init__(v_type, brand)
        self.battery = battery


my_ev = ElectricCar("Sedan", "Tesla", "85 kWh")
print(f"Type: {my_ev.type}, Brand: {my_ev.brand}, Battery: {my_ev.battery}")
```

```python
#ANSWER 4
class Bird:
    def fly(self):
        print("Most birds can fly.")

class Sparrow(Bird):
    def fly(self):
        print("Sparrows fly high in the sky.")

class Penguin(Bird):
    def fly(self):
        print("Penguins cannot fly, but they can swim!")

birds = [Sparrow(), Penguin()]
for bird in birds:
    bird.fly()
```

```python
#ANSWER 5
class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New Balance: ${self.__balance}")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew ${amount}. New Balance: ${self.__balance}")
        else:
            print("Insufficient funds or invalid amount.")

    def check_balance(self):
        return self.__balance

account = BankAccount(100)
account.deposit(50)
account.withdraw(30)
print(f"Final Balance: ${account.check_balance()}")
```

```python
#ANSWER 6
class Instrument:
    def play(self):
        print("Playing a generic instrument.")

class Guitar(Instrument):
    def play(self):
        print("Strumming the guitar strings.")

class Piano(Instrument):
    def play(self):
        print("Playing a classical melody on the piano.")

def performance(instrument_instance):
    instrument_instance.play()

performance(Guitar())
performance(Piano())
```

```python
#ANSWER 7
class MathOperations:
    @classmethod
    def add_numbers(cls, a, b):
        return a + b

    @staticmethod
    def subtract_numbers(a, b):
        return a - b


print("Addition via classmethod:", MathOperations.add_numbers(10, 5))
print("Subtraction via staticmethod:", MathOperations.subtract_numbers(10, 5))
```

```python
#ANSWER12
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average_grade(self):
        if not self.grades:
            return 0
        return sum(self.grades) / len(self.grades)

student = Student("Sam", [85, 90, 78, 92])
print(f"{student.name}'s Average Grade: {student.average_grade():.2f}")
```

```python
#answer14
class Employee:
    def __init__(self, name, hours_worked, hourly_rate):
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class Manager(Employee):
    def __init__(self, name, hours_worked, hourly_rate, bonus):
        super().__init__(name, hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        base_salary = super().calculate_salary()
        return base_salary + self.bonus

emp = Employee("John", 40, 25)
mgr = Manager("Sarah", 40, 50, 500)
print(f"Employee Salary: ${emp.calculate_salary()}")
print(f"Manager Salary: ${mgr.calculate_salary()}")
```

```python
#ANSWER15
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_price(self):
        return self.price * self.quantity

prod = Product("Laptop", 1200, 3)
print(f"Total inventory cost for {prod.name}: ${prod.total_price()}")
```

```python
#ANSWER16
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        return "Moo"

class Sheep(Animal):
    def sound(self):
        return "Baa"

animals = [Cow(), Sheep()]
for a in animals:
    print(f"{type(a).__name__} says: {a.sound()}")
```

```python
#ANSWER17
class Book:
    def __init__(self, title, author, year_published):
        self.title = title
        self.author = author
        self.year_published = year_published

    def get_book_info(self):
        return f"Title: '{self.title}', Author: {self.author} ({self.year_published})"

book = Book("1984", "George Orwell", 1949)
print(book.get_book_info())
```

```python
#ANSWER18
class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

class Mansion(House):
    def __init__(self, address, price, number_of_rooms):
        super().__init__(address, price)
        self.number_of_rooms = number_of_rooms

luxury_home = Mansion("123 Beverly Hills Dr", 5000000, 14)
print(f"Mansion details -> Address: {luxury_home.address}, Price: ${luxury_home.price}, Rooms: {luxury_home.number_of_rooms}")
```
